## Google Colab Keyboard Shortcuts

| Action | Windows/Linux | macOS |
| :--- | :--- | :--- |
| **Run cell** | `Ctrl + Enter` | `Command + Enter` |
| **Run cell and select next** | `Shift + Enter` | `Shift + Enter` |
| **Run cell and insert below** | `Alt + Enter` | `Option + Enter` |
| **Run all cells** | `Ctrl + F9` | `Command + F9` |
| **Deselect cell** | `Esc` | `Esc` |
| **Move to previous cell (Up)** | `Ctrl + M + P` or `K` | `Command + M + P` or `K` |
| **Move to next cell (Down)** | `Ctrl + M + N` or `J` | `Command + M + N` or `J` |
| **Move selected cell up** | `Ctrl + M + K` | `Command + M + K` |
| **Move selected cell down** | `Ctrl + M + J` | `Command + M + J` |
| **Insert code cell above** | `Ctrl + M + A` | `Command + M + A` |
| **Insert code cell below** | `Ctrl + M + B` | `Command + M + B` |
| **Delete cell** | `Ctrl + M + D` | `Command + M + D` |
| **Convert to Markdown** | `Ctrl + M + M` | `Command + M + M` |
| **Convert to Code** | `Ctrl + M + Y` | `Command + M + Y` |
| **Find and replace** | `Ctrl + H` | `Ctrl + H` |
