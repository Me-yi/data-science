## `pandas.MultiIndex`
(_2026-07-04T10:56:50+07:00 - AI Generated with Gemini 3.5 Flash (Low)_)

A **`pandas.MultiIndex`** (also known as a hierarchical index) allows you to have multiple index levels on an axis (either rows or columns) of a pandas `DataFrame` or `Series`.

This enables you to represent and manipulate higher-dimensional data in a standard two-dimensional tabular format.

### Key Concepts

1. **Hierarchical Structure**: Instead of a flat list of labels (like `['A', 'B', 'C']`), a `MultiIndex` consists of nested tuples of labels (like `[('Group 1', 'A'), ('Group 1', 'B'), ('Group 2', 'A')]`).
2. **Levels and Codes**:
    - **`levels`**: The unique labels at each category level.
    - **`codes`**: Integer arrays indicating which label from each level is present at each position.
3. **Dimensions**: It lets you work with data that has three or more dimensions using standard 2D `DataFrame` operations.

## Basic Example

Here is how you can create and use a `MultiIndex` `DataFrame`:

````python
# Define arrays for the levels
outer_index = ['State A', 'State A', 'State B', 'State B']
inner_index = ['City 1', 'City 2', 'City 1', 'City 2']

# Create a MultiIndex from the arrays
index = pd.MultiIndex.from_arrays(
    [outer_index, inner_index],
    names=['State', 'City']
)
index
````

````python
# Create a DataFrame using the MultiIndex
df_multiIndex = pd.DataFrame({
    'Population (M)': [1.2, 0.8, 3.4, 2.1]
}, index=index)
df_multiIndex
````

### Common Use Cases

#### Grouping/Aggregating

When you perform a `groupby()` operation on multiple columns, pandas automatically returns a `MultiIndex` object:

````python

arrays = [
    ["US", "US", "US", "JP", "JP", "JP"],
    ["2025", "2025", "2026", "2025", "2026", "2026"],
    ["Q1", "Q2", "Q1", "Q1", "Q1", "Q2"]
]

index = pd.MultiIndex.from_arrays(arrays, names=["country", "year", "quarter"])

df_country_quarter = pd.DataFrame({
    "sales": [100, 120, 150, 80, 90, 110],
    "expenses": [60, 70, 80, 50, 55, 65]
}, index=index)
df_country_quarter
````

````python
# Group by the 'country' level using its name
country_totals = df_country_quarter.groupby(level="country").sum()
country_totals
````

````python
# Alternatively, group by the outermost level using its integer index (0)
country_totals = df_country_quarter.groupby(level=0).sum()
country_totals
````

**Reshaping**: Operations like `.stack()` (moving column labels to row index) and `.unstack()` (moving row index levels to column labels) produce `MultiIndex` structures.
- **Complex Slicing**: You can select data at specific levels using `.xs()` (cross-section) or by passing tuples to `.loc[]`.


## `pandas.DataFrame.apply`

(2026-07-05T10:52:10+07:00 - AI Generated with Gemini 3.5 Flash (Medium))

`pandas.DataFrame.apply` allows you to apply a function along a specific axis of a DataFrame (either column-by-column or row-by-row).

```
       Axis 1 (columns) ───>
         Col A   Col B   Col C
       ┌───────┬───────┬───────┐
Row 0  │   1   │  10   │  100  │  │
Row 1  │   2   │  20   │  200  │  ▼ Axis 0 (index/rows)
Row 2  │   3   │  30   │  300  │
       └───────┴───────┴───────┘
```

It is highly flexible and useful for executing custom logic that isn't built into pandas' standard vectorized methods.

|Axis Number|Representation|Action Taken|Input for Function|Common Use|
|--|--|--|--|--|
|`axis=0`|Table index|pandas takes one column at a time, grabs all the row values for that column, and sends them to your function as a single unit (a pandas.Series)|The function receives Column A, then Column B, then Column C| Calculating a summary statistic per column (e.g., column sums, column means, column ranges)|
|`axis=1`|Table columns|pandas takes one row at a time, grabs all the column values for that row, and sends them to your function as a single unit (a pandas.Series)|The function receives Row 0, then Row 1, then Row 2| Combining values from different columns in the same row (e.g., adding Column A and Column B for each row)|

````python
apply_example_df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [10, 20, 30],
    'C': [100, 200, 300]
})
````

### Applying to Columns (axis=0)

````python
# The custom function receives a Series (a column)
def find_range(col):
    return col.max() - col.min()
column_ranges = apply_example_df.apply(find_range, axis=0)
column_ranges
````

### Applying to Rows (axis=1)

````python
# The custom function receives a Series (a row)
def calculate_custom_score(row):
    return row['A'] + row['B'] - row['C']

apply_example_df['Score'] = apply_example_df.apply(calculate_custom_score, axis=1)
apply_example_df
````

### Using Lambda Functions and Passing Arguments

You can use lambda for quick, one-line functions and pass extra arguments using the args tuple or keyword arguments:

````python
# Add a scalar offset to all values using apply
df_offset = apply_example_df.apply(lambda val, offset: val + offset, args=(5,))
df_offset
````

### Key Parameters to Know

- `raw` (bool, default `False`): If `True`, the function receives `numpy.ndarray` objects instead of `pandas.Series` objects. This provides a significant speedup if you do not need index labels inside your custom function.
- `result_type` (for `axis=1` only):
  - `'expand'`: Returns a DataFrame where list-like results will be turned into separate columns.
  - `'reduce'`: Returns a Series if possible, rather than expanding list-like results into a DataFrame.
  - `'broadcast'`: Retains the original DataFrame shape and index, broadcasting the result across the columns.

### Performance
While apply is extremely versatile, it behaves like a standard Python loop under the hood. Consequently, it can be slow on large datasets.

- __Vectorized Operations__: Whenever possible, use pandas' native vectorized operations, which are implemented in optimized C.
  - _Slow (using apply)_: `df['A'].apply(lambda x: x * 2)`
  - _Fast (vectorized)_: `df['A'] * 2`
- __Numpy Functions__: Many NumPy functions can be passed directly to `apply` and run very efficiently:
  - `df.apply(np.sqrt)`

## pandas.DataFrame.crosstab

(_2026-07-04T12:00:24+07:00 - AI Generated with Gemini 3.5 Flash (Low)_)

The **`pandas.crosstab()`** function computes a simple cross-tabulation (contingency table) of two or more factors. By default, it computes a frequency table of the factors (how many times each combination occurs), but it can also aggregate other data using an aggregation function.

### Basic Usage: Frequency Table
Here is a basic example showing how `crosstab` counts the occurrences of combinations:

````python
# Sample dataset of store sales transactions
stores_products_df = pd.DataFrame({
    'Store': ['North', 'North', 'South', 'North', 'South', 'South'],
    'Product': ['Apple', 'Banana', 'Apple', 'Apple', 'Banana', 'Banana']
})

# Count how many times each product is sold at each store
ct = pd.crosstab(stores_products_df['Store'], stores_products_df['Product'])
print(ct)
````

## pandas.DataFrame.describe
(_2026-07-04T12:08:08+07:00 - AI Generated with Gemini 3.5 Flash (Low)_)

The **`describe()`** method in pandas is used to generate descriptive statistics that summarise the central tendency, dispersion, and shape of a dataset's distribution, excluding `NaN` values.

### Numeric Data Behaviour
By default, calling `describe()` on a `DataFrame` containing numeric data returns basic statistical metrics for all numeric columns:

````python
salaries_age_df = pd.DataFrame({
    'Age': [25, 47, 31, 22, 35, 29],
    'Salary': [50000, 120000, 71000, 48000, 85000, 62000]
})

print(salaries_age_df.describe())
````

- **`count`**: Number of non-null values.
- **`mean`**: The arithmetic average.
- **`std`**: Standard deviation (dispersion).
- **`min` / `max`**: Minimum and maximum values.
- **`25%`, `50%`, `75%`**: The percentiles (50% is the median).

### Non-Numeric (Categorical/String) Data Behaviour

If you run `describe()` on object or categorical columns, it returns descriptive metrics specific to categorical variables:

````python
df_cat = pd.DataFrame({
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'IT', 'HR']
})

print(df_cat.describe())
````

- **`unique`**: The number of distinct categories.
- **`top`**: The most common category (mode).
- **`freq`**: The frequency of the most common category.

## Key Parameters
- **`include`**: Specify what columns to include in the output.
    - `df.describe(include='all')`: Summarizes both numeric and categorical columns together.
    - `df.describe(include=['O'])`: Summarizes only string/object columns.
- **`percentiles`**: Modify the returned percentiles (defaults to `[.25, .5, .75]`).
    - `df.describe(percentiles=[.1, .5, .9])`

## `pandas.DataFrame.value_counts`
(_2026-07-06T11:16:21+07:00 - AI Generated with Gemini 3.5 Flash (Medium)_)

In Pandas, `value_counts()` is one of the most frequently used functions for data exploration. It counts the occurrences of unique values in a dataset.

While historically accessed as `pd.value_counts(series)`, it is now standard practice to call it directly as a method on a **Series** or a **DataFrame**, as the top-level `pd.value_counts()` function was deprecated in Pandas 2.0.

### Basic Usage on a Series
When called on a Series, it returns a new Series where the index contains the unique values from the original Series, and the values contain their corresponding counts. By default, the result is sorted in descending order.

````python
# Sample Series
s = pd.Series(['apple', 'banana', 'apple', 'orange', 'banana', 'apple'])

# Get counts
print(s.value_counts())
````

### Series - Key Parameters

The `value_counts()` method has several useful parameters:

#### normalize
- **`normalize`** _(bool, default False)_: If set to `True`, it returns the relative frequencies (proportions) of the unique values instead of the raw counts.

````python
s.value_counts(normalize=True)
````

#### dropna

**`dropna`** _(bool, default True)_: By default, missing values (`NaN`) are excluded from the counts. Set `dropna=False` to include them.

````python
s_with_nan = pd.Series(['apple', None, 'apple', 'orange'])
s_with_nan.value_counts(dropna=False)
````

#### ascending

**`ascending`** _(bool, default False)_: By default, values are sorted in descending order (highest count first). Set `ascending=True` to sort in ascending order.

````python
s.value_counts(ascending=True)
````

#### bins

**`bins`** _(int, optional)_: Instead of counting unique discrete values, you can group continuous numeric data into half-open bins. This is a convenient shortcut for `pd.cut()`

````python
numbers = pd.Series([1, 2, 2, 3, 10, 12, 15])
numbers.value_counts(bins=3)
````

### Usage on a DataFrame

You can also run `value_counts()` on a DataFrame. This returns the counts of unique _rows_ (combinations of values across columns) in the DataFrame.

````python
df_fruit_colors = pd.DataFrame({
    'fruit': ['apple', 'banana', 'apple', 'apple'],
    'color': ['red', 'yellow', 'red', 'green']
})
df_fruit_colors.value_counts()
````

## `pandas.DataFrame.isna`
(_2026-07-04T12:18:14+07:00 - AI Generated with Gemini 3.5 Flash (Low)_)

The **`isna()`** method in pandas is used to detect missing or null values in a `DataFrame` or `Series`. It returns a boolean object of the same shape, where `True` indicates a missing value (`NaN`, `None`, or `NaT`) and `False` indicates a valid value.

> [NOTE!]
> `isna()` and `isnull()` are completely identical helper functions that perform the exact same operation.

````python
df_with_isna = pd.DataFrame({
    'Name': ['Alice', 'Bob', None],
    'Age': [25, np.nan, 30]
})

df_with_isna.isna()
````

### Common Operations with `isna()`

#### Counting Missing Values per Column

You can chain `.sum()` to count the total number of missing values in each column:

````python
df_with_isna.isna().sum()
````

#### Filtering Rows with Missing Values

You can use the boolean mask returned by `isna()` to filter rows containing missing values:

````python
# Filter rows where 'Age' is missing
missing_age = df_with_isna[df_with_isna['Age'].isna()]
missing_age
````

### Filtering Rows with NO Missing Values

You can use the tilde operator `~` to invert the mask and select non-null values:

````python
# Keep rows where 'Age' is not null
valid_age = df_with_isna[~df_with_isna['Age'].isna()]
valid_age
````
