# Transformations

## Required imports

```python
from pathlib import Path

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Data sources

### Get DataFrame

```python
def get_df(file_name: str):
  filepath = Path("..")/"data"/file_name
  return pd.read_csv(filepath, index_col=0)
```

### Sources

```python
world_happiness = get_df('world_happiness.csv')
```

## `np.log()`

```python
# Create log_gdp_per_cap column
world_happiness['log_gdp_per_cap'] = np.log(world_happiness['gdp_per_cap'])

# Scatterplot of happiness_score vs. log_gdp_per_cap
sns.scatterplot(data=world_happiness, x='log_gdp_per_cap', y='happiness_score')
plt.show()
```

```python
# Calculate correlation
corr = world_happiness['log_gdp_per_cap'].corr(world_happiness['happiness_score'])
corr
```
