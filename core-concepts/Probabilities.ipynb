## What is probability

(_2026-04-18T11:02:35+01:00Z - AI Generated with Gemini 3 Flash - Thinking_)

At its simplest, **probability** is the branch of mathematics that quantifies uncertainty. It measures how likely an event is to occur, represented as a value between **0** and **1** (or 0% to 100%).


## Calculating probabilities


### What's the probability of an event?

$$P(\text{event}) = \frac{\text{\# ways event can happen}}{\text{total \# of possible outcomes}}$$

## Getting Samples

### Without replacement

```python
# Including the random_state, localises the random seed
amir_deals.sample(5, random_state=42)
```

### With replacement

```python
# Including the random_state, localises the random seed
amir_deals.sample(5, random_state=42, replace=True)
```

## Independent events

Two events are **independent** if the probability of the second event **isn't** affected by the outcome of the first event.

Where **no** replacement sampling has taken place.

## Dependent events

Two events are **dependent** if the probability of the second event **is** affected by the outcome of the first event.

Where replacement sampling has taken place.

## Calculating probabilities - product types

```python
# Count the deals for each product
counts = amir_deals['product'].value_counts()
counts
```

```python
# Calculate probability of picking a deal with each product
probs = counts.apply(lambda x: x/int(counts.sum()))
probs
```
